```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("akshaydattatraykhare/diabetes-dataset")

print("Path to dataset files:", path)
```

```python
import os

os.listdir(path)

import pandas as pd

import numpy as np

df = pd.read_csv(path+'/diabetes.csv')
```

```python
df.head()
```

```python
min = {}
max = {}

for col in df.columns:
    max[col] = df[col].max()
    min[col] = df[col].min()
```

```python
df.columns
```

```python
newdf = {}

for col in df.columns:
    newdf[col] = []
```

```python
for row in df.iterrows():
    for col in df.columns:
        row[1][col] = (max[col] - row[1][col])/(max[col]-min[col])
        newdf[col].append(row[1][col])
    # print(row)

    # break
```

```python
newdf
```

```python
newds = pd.DataFrame(newdf)
```

```python
newds.head()
```

```python
len(newds)
```

```python
x_train = newds[0:int(len(newds)*0.9)]
x_test = newds[int(len(newds)*0.9):-1]
```

```python
x_train.head()
```

```python
from sklearn.linear_model import LogisticRegression

X = ['Pregnancies', 'Glucose', 'BloodPressure', 'SkinThickness', 'Insulin',
       'BMI', 'DiabetesPedigreeFunction', 'Age']
y = ['Outcome']

model = LogisticRegression()

model.fit(x_train[X], x_train[y])

```

```python
pred = model.predict(x_test[X])
```

```python
from sklearn.metrics import accuracy_score

print(accuracy_score(pred, x_test[y]))
```

```python
import pickle

filename = 'diabetes_model.pkl'
# Load the model from disk
with open(filename, 'wb') as file:
    pickle.dump(model, file)
```
